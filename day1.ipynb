```python
from openai import OpenAI
```

```python
openai = OpenAI()
```

```python
query = "Greatest UFC fighter of all time"

sending_prompt = [{"role": "user", "content": query}]
```

```python
response = openai.chat.completions.create(model = "gpt-4o-mini",messages = sending_prompt)
print(response.choices[0].message.content)
```

```python
prompt = "Give a buisness idea prompt i can input in an ai,only prompt nothing else"
sending_prompt = [{"role":"user", "content" : prompt }]
```

```python
response = openai.chat.completions.create(model = "gpt-4o-mini",messages = sending_prompt)
prompt_for_ai = response.choices[0].message.content
print(prompt_for_ai)
```

```python
query = [{"role":"user", "content":prompt_for_ai}]
response = openai.chat.completions.create(model = "gpt-4o-mini",messages = query)
idea = response.choices[0].message.content
print(idea)
```
