```python
from openai import OpenAI
```

```python
openai = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"  
)
```

```python
suggestions = "Suggest me 10 great projects that i can do in Agentic AI domain"
messages = [{"role":"user","content":suggestions}]
print(messages)
```

```python
response = openai.chat.completions.create(model="qwen3:8b", messages=messages)
suggestions = response.choices[0].message.content
print(suggestions)
```

```python
choose = f"""Just selet 1 which you think would be good for college project from {suggestions} and just write about it """
query = [{"role":"user","content":choose}]
```

```python
response = openai.chat.completions.create(model="qwen3:8b",messages=query)
selected_project = response.choices[0].message.content
print(selected_project)
```

```python
project = f"""Just give a report for how to do this {selected_project}"""
messages = [{"role":"user","content":project}]
response = openai.chat.completions.create(model="qwen3:8b",messages=messages)
print(response.choices[0].message.content)
```
